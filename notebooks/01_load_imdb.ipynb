# Step 1 & 2: Loading and Exploring the IMDb Large Movie Review Dataset

**Project:** Aspect-Based Sentiment Analysis (ABSA) on Movies  
**Objective:** Fine-tune a transformer model to extract audience sentiment regarding specific actors and plot points directly from unstructured theater/movie reviews.  

---
### Dataset Overview
- **Source:** `stanfordnlp/imdb` via Hugging Face Datasets
- **Train Set:** 25,000 labeled reviews (12,500 positive, 12,500 negative)
- **Test Set:** 25,000 labeled reviews (12,500 positive, 12,500 negative)
- **Unsupervised:** 50,000 unlabeled reviews
- **Key Reality:** The IMDb dataset contains *document-level* binary sentiment labels (0 = Negative, 1 = Positive). It does **not** natively contain token-level aspect annotations. Our pipeline will extract aspects (actors, plot, direction, etc.) and infer aspect-level sentiment.
